```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
task1 = pd.read_csv(r'Task 1.csv')
task1_50 = pd.read_csv(r'Task 1 50 Ohm.csv')

task1
```

```python
frequency_1 = task1['Frequency (Hz)']
re_S11_1 = task1['Re(S11)']
im_S11_1 = task1['Im(S11)']

re_S21_1 = task1['Re(S21)']
im_S21_1 = task1['Im(S21)']

frequency_1_50 = task1_50['Frequency (Hz)']
re_S11_1_50 = task1_50['Re(S11)']
im_S11_1_50 = task1_50['Im(S11)']

re_S21_1_50 = task1_50['Re(S21)']
im_S21_1_50 = task1_50['Im(S21)']
```

```python
LogMag_S11 = 10 * np.log10(re_S11_1**2 + im_S11_1**2)
LogMag_S21 = 10 * np.log10(re_S21_1**2 + im_S21_1**2)

LogMag_S11_50 = 10 * np.log10(re_S11_1_50**2 + im_S11_1_50**2)
LogMag_S21_50 = 10 * np.log10(re_S21_1_50**2 + im_S21_1_50**2)
```

```python
plt.title("Power vs. Frequency Task 1 - Initial Copper Width 6mm")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Power (dBm)")
plt.scatter(frequency_1, LogMag_S11, label='S11')
plt.scatter(frequency_1, LogMag_S21, label='S21')
plt.legend()
plt.show()

plt.title("Power vs. Frequency Task 1 - Copper Width 2.7mm")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Power (dBm)")
plt.scatter(frequency_1_50, LogMag_S11_50, label='S11')
plt.scatter(frequency_1_50, LogMag_S21_50, label='S21')
plt.legend()
plt.show()
```
